Q10: Mini Project – Simple Calculator
Write a Python program that:
● Takes two numbers from the user.
● Performs +, -, *, /, %, and **.
● Displays the results clearly.
● Uses appropriate variables, input, type conversion, and arithmetic operators.

```python
num1= int(input("Enter the nuumber "))
num2= int(input("Enter the another number"))
Addition = num1 + num2
print("Addition of numbers is ->", Addition)
Sub = num1 - num2
print("Subtraction of numbers is ->", Sub)
Multi= num1 * num2
print("Multipliction of numbers is ->", Multi)
Div = num1 / num2
print("Division of numbers is ->", Div)
Rem = num1 % num2
print("Remainder of numbers is ->", Rem)
power = num1 ** num2
print("Power of numbers is ->", power)
```
